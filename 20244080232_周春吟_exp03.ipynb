```python
import re
from bs4 import BeautifulSoup
```

```python
text1 = "cat on mat; later: cats and dogs"

res_d = re.findall(r"\d+", text1)
res_s = re.findall(r"\s+", text1)
res_cat_word = re.findall(r"\bcat\b", text1)

print("\\d+ 匹配结果：", res_d)
print("\\s+ 匹配结果：", res_s)
print(r"\bcat\b 匹配结果：", res_cat_word)
```

```python
text2 = "cat dog bird CAT"
res_cd = re.findall(r"cat|dog", text2)
print("cat|dog匹配：", res_cd)

r1 = re.findall(r"a|aa|aaa", "aaaa")
r2 = re.findall(r"aaa|aa|a", "aaaa")
print("r'a|aa|aaa'结果：", r1)
print("r'aaa|aa|a'结果：", r2)
```

```python
text3 = "ct cat caat caaat"

p1 = re.findall(r"ca*t", text3)
p2 = re.findall(r"ca+t", text3)
p3 = re.findall(r"ca?t", text3)
p4 = re.findall(r"ca{2}t", text3)
p5 = re.findall(r"ca{1,2}t", text3)

print("ca*t  :", p1)
print("ca+t  :", p2)
print("ca?t  :", p3)
print("ca{2}t:", p4)
print("ca{1,2}t:", p5)
```

```python
mail_text = "联系:alice@example.com;备选:bob@test.org"
# 捕获组：(用户名)@(域名)
mail_pattern = r"([\w.+-]+)@([\w-]+\.[a-zA-Z0-9.]+)"
mails = re.findall(mail_pattern, mail_text)
print("(用户名,域名):")
for user, domain in mails:
    print(f"用户名：{user}，域名：{domain}")
```

```python
dup_text = "cat cat, dog cat, bird bird"
# \1代表引用第1捕获组匹配到的单词
dup_pattern = r"(\b\w+\b)\s+\1"
repeat_words = re.findall(dup_pattern, dup_text)
print("连续重复出现的单词：", repeat_words)
```

```python
date_text = """2024-05-01
2024/05/02
2024.05.03"""
print("====转换前====")
print(date_text)

# 分隔符可以是 - / .
date_pat = r"(\d{4})([-/.])(\d{2})([-/.])(\d{2})"
date_out = re.sub(date_pat, r"\3-\5-\1", date_text)

print("\n====转换后====")
print(date_out)
```

```python
phone_text = "联系我:123-456-7890 或 987.654.3210"
phone_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
phone_result = re.sub(phone_pat, r"(\1) \2-\3", phone_text)
print("转换后电话号码文本：")
print(phone_result)
```

```python
html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")

# 任务1
title_text = soup.title.get_text()
h1_tag = soup.h1
print("页面标题文本：", title_text)
print("h1标签名：", h1_tag.name)
print("h1文本内容：", h1_tag.get_text())
```

```python
# 所有class="course"段落
p_list = soup.find_all("p", class_="course")
print("\n===class='course'的p段落===")
for p in p_list:
    print(p.get_text())

# 全部a链接
a_list = soup.find_all("a")
print("\n===全部链接===")
for a in a_list:
    link_text = a.get_text()
    href_val = a["href"]
    print(f"链接文字:{link_text}, href:{href_val}")
```
