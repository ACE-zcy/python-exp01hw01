```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

#测试三组
print("gcd(2018,2019) =", gcd(2018,2019))
print("gcd(1200,300) =", gcd(1200,300))
print("gcd(5040,60) =", gcd(5040,60))

#测试负数
print("\n负数测试 gcd(-12, -8):",gcd(-12,-8))
```

```python
import math

#(1) euler_limit (1+1/n)^n
def euler_limit(n):
    return (1 + 1/n)**n

#(2) euler_infinite_sum 前n项求和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(0,n):
        if k>0:
            fact *= k
        total += 1.0 / fact
    return total

#(3) euler_approx 根据误差epsilon逼近e
def euler_approx(epsilon):
    real_e = math.exp(1)
    k = 0
    while True:
        s = euler_infinite_sum(k)
        if abs(s - real_e) < epsilon:
            return s
        k +=1

#(4) print_euler_sum_table
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k}, sum={val:.6f}")


#测试
print("euler_limit(1000):", euler_limit(1000))
print("euler_infinite_sum(4):", euler_infinite_sum(4))
print("真实e =", math.exp(1))
print("epsilon=0.001近似e：", euler_approx(0.001))
print("\n==== sum table ====")
print_euler_sum_table(6)
```

```python
# is_power_of_2 判断是否是2的幂
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n //2
    return True

# is_power(b,n) n是否为b的幂
def is_power(b,n):
    while n>1:
        if n % b !=0:
            return False
        n = n // b
    return True


#测试
print("is_power_of_2(8):", is_power_of_2(8))
print("is_power_of_2(7):", is_power_of_2(7))

print("\nis_power(3,81):",is_power(3,81))
print("is_power(2,10):",is_power(2,10))
```
