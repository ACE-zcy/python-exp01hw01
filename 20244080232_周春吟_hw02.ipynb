```python
# ===== 1、字符串的乐趣 =====

# 1.1 回文判断：忽略空格和大小写
def is_palindrome(s):
    s = s.replace(" ", "").lower()
    return s == s[::-1]

# 1.2 字母顺序判断：忽略空格
def is_abecedarian(s):
    s = s.replace(" ", "").lower()
    return all(s[i] <= s[i+1] for i in range(len(s)-1))

# 1.3 删除元音 a e i o u
def remove_vowels(s):
    vowels = set("aeiouAEIOU")
    return "".join(c for c in s if c not in vowels)

# ---- 测试 ----
print("== 回文 ==")
print(is_palindrome("tacocat"))          # True
print(is_palindrome("TACO cat"))         # True
print(is_palindrome("rats live on no evil star"))  # True
print(is_palindrome("hello"))            # False

print("\n== 字母顺序 ==")
print(is_abecedarian("adder"))           # True
print(is_abecedarian("beet"))            # True
print(is_abecedarian("abcd efgh xyz"))   # True
print(is_abecedarian("dog"))             # False

print("\n== 删除元音 ==")
print(remove_vowels("cat"))              # ct
print(remove_vowels("audio"))            # d
print(remove_vowels("aeiou"))            # （空字符串）
```

```python
# ===== 2、列表的乐趣 =====

# 2.1 列表反转，返回新列表，不修改原列表
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表")
    return lst[::-1]

# 2.2 升序判断，只遍历一次
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

# ---- 测试 ----
print("== 列表反转 ==")
print(list_reverse([1, 2, 3]))           # [3, 2, 1]
a = [1, 2, 3]
print("原列表不变：", a)                  # [1, 2, 3]
try:
    list_reverse("abc")
except TypeError as e:
    print("抛出 TypeError：", e)

print("\n== 升序判断 ==")
print(is_sorted([1, 2, 3, 4]))           # True
print(is_sorted([1, 3, 2]))              # False
print(is_sorted([4, 4, 5]))              # True
print(is_sorted((1, 2, 3)))              # True（元组也可以）
```

## 问题2：列表的乐趣

实现列表反转与升序判断两个函数。
- `list_reverse`：输入不是列表时抛出错误。
- `is_sorted`：只遍历序列一次，升序返回 True。


```python
# ===== 3、作为向量的元组 =====

# 3.1 标量乘法
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("第二个参数必须是整数或浮点数")
    return tuple(x * s for x in t)

# 3.2 内积（返回浮点数）
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(x) != len(y):
        raise ValueError("两个元组长度不一致")
    return float(sum(a * b for a, b in zip(x, y)))

# 3.3 有效矩阵检查
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    for row in mx:
        if not isinstance(row, tuple):
            return False
    # 所有内层元组长度相等
    if len(mx) > 0:
        n = len(mx[0])
        for row in mx:
            if len(row) != n:
                return False
    # 每个元素必须是整数或浮点数
    for row in mx:
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

# 3.4 矩阵 × 向量，返回数值元组
def mx_vec_mult(mx, v):
    if not check_valid_mx(mx):
        raise ValueError("第一个参数不是有效矩阵")
    if not isinstance(v, tuple):
        raise TypeError("第二个参数必须是元组")
    if len(mx) == 0:
        return ()
    if len(v) != len(mx[0]):
        raise ValueError("向量长度与矩阵列数不一致")
    result = []
    for row in mx:
        result.append(sum(a * b for a, b in zip(row, v)))
    return tuple(result)

# ---- 测试 ----
print("== 标量乘法 ==")
print(vec_scalar_mult((1, 2, 3), 2))     # (2, 4, 6)
print(vec_scalar_mult((1.5, 2.5), 3))    # (4.5, 7.5)

print("\n== 内积 ==")
print(vec_inner_product((1, 2), (3, 4))) # 11.0
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("抛出 ValueError：", e)

print("\n== 矩阵检查 ==")
print(check_valid_mx(((1, 2), (3, 4))))              # True
print(check_valid_mx(((1, 2), (3,))))                 # False 长度不一致
print(check_valid_mx(((1, "a"), (2, 3))))             # False 含非数字
print(check_valid_mx([1, 2]))                         # False 外层不是元组

print("\n== 矩阵乘向量 ==")
print(mx_vec_mult(((1, 2), (3, 4)), (1, 1)))          # (3, 7)
print(mx_vec_mult(((1, 0, 0), (0, 1, 0)), (2, 3, 4))) # (2, 3)
```

## 问题3：作为向量的元组

用元组表示向量：长度为 d 的元组表示 d 维向量。
实现标量乘法、内积、矩阵合法性检查、矩阵乘向量。
需对输入类型做检查并抛出合适的 TypeError / ValueError。


```python
# ===== 4、Vector 类 =====
class Vector:
    def __init__(self, dim, elements=None):
        # 维度必须是非负整数
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        # 只给维度时，默认元素全为 0
        if elements is None:
            elements = [0.0] * dim
        if not isinstance(elements, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(elements) != dim:
            raise ValueError("元素个数与维度不一致")
        for e in elements:
            if not isinstance(e, (int, float)):
                raise TypeError("元素必须是整数或浮点数")
        self.dim = dim
        self.elements = list(elements)

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # 维度不同的向量比较时抛异常
    def _check_compare(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与 Vector 对象比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        self._check_compare(other)
        return self.elements == other.elements

    def __lt__(self, other):
        self._check_compare(other)
        return self.elements < other.elements

    def __le__(self, other):
        self._check_compare(other)
        return self.elements <= other.elements

    def __gt__(self, other):
        self._check_compare(other)
        return self.elements > other.elements

    def __ge__(self, other):
        self._check_compare(other)
        return self.elements >= other.elements

    # 内积
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("两个向量维度不一致")
        return sum(a * b for a, b in zip(self.elements, other.elements))

    # 乘法：标量 or 逐元素 Vector
    def __mul__(self, other):
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("两个向量维度不一致")
            return Vector(self.dim, [a * b for a, b in zip(self.elements, other.elements)])
        if isinstance(other, (int, float)):
            return Vector(self.dim, [a * other for a in self.elements])
        return NotImplemented

    # 标量左乘：2 * v
    def __rmul__(self, other):
        if isinstance(other, (int, float)):
            return Vector(self.dim, [a * other for a in self.elements])
        return NotImplemented

# ---- 测试 ----
v1 = Vector(2, (2, 3))
v2 = Vector(2, (2, 4))
v3 = Vector(2, (2, 3))

print("== 构造与默认 ==")
print(v1)
print(Vector(3))                      # 默认元素全 0
try:
    Vector(-1, (1, 2))
except ValueError as e:
    print("抛 ValueError：", e)
try:
    Vector(2, (1,))
except ValueError as e:
    print("抛 ValueError：", e)

print("\n== 比较 ==")
print("v1 == v3 :", v1 == v3)         # True
print("v1 == v2 :", v1 == v2)         # False
print("v1 < v2  :", v1 < v2)          # True
print("v2 > v1  :", v2 > v1)          # True
print("v1 <= v3 :", v1 <= v3)         # True

print("\n== 内积 ==")
print("v1.dot(v2) =", v1.dot(v2))     # 2*2 + 3*4 = 16

print("\n== 乘法 ==")
print("2 * v1 =", 2 * v1)             # Vector(2, [4, 6])
print("v1 * 3 =", v1 * 3)             # Vector(2, [6, 9])
print("v1 * v2 =", v1 * v2)           # Vector(2, [4, 12])

print("\n== 维度不同的向量比较会报错 ==")
try:
    Vector(2, (1, 2)) < Vector(3, (1, 2, 3))
except ValueError as e:
    print("抛 ValueError：", e)
```

## 问题4：更多向量的乐趣（Vector 类）

用面向对象方式实现向量：
- 构造函数接收维度与元素，只给维度时默认元素全 0；
- 维度与元素个数不一致、维度为负或类型错误时抛出异常；
- 实现比较运算符（按坐标字典序）与 `dot` 内积、标量/逐元素乘法。

### 维度和元素是实例属性还是类属性？

应是**实例属性**。理由：
- 每个 Vector 对象代表一个独立的向量，拥有各自不同的维度和元素值；
- 若用类属性，所有对象会共享同一份维度与元素，无法区分不同向量，也不符合"每个向量对象有自己的状态"的语义；
- 实例属性存储在每个对象自己的命名空间中，互不干扰，符合面向对象"对象封装自身状态"的设计原则。
