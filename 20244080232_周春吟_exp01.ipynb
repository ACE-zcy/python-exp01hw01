```python
# ========== 实验2：函数、条件、循环与递归 ==========
# 1. 分数等级函数
def score_level(score):
    if not isinstance(score, (int, float)):
        raise ValueError("分数必须是数字")
    if score < 0 or score > 100:
        raise ValueError("分数范围必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:  # 0 <= score <60
        return "不及格"

print("===== 分数等级函数测试 =====")
test_scores = [95,82,74,66,50]
for s in test_scores:
    print(f"分数{s} → {score_level(s)}")

# 2. for循环求1~100累加和
sum_for = 0
for i in range(1,101):
    sum_for += i
print(f"\n===== for循环 1-100求和：{sum_for}")

# 3. while循环求1~100累加和
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print(f"===== while循环 1-100求和：{sum_while}")

# 4. 迭代版阶乘
def fact_iter(n):
    if not isinstance(n,int) or n < 0:
        raise ValueError("n必须是非负整数")
    res = 1
    for i in range(1, n+1):
        res *= i
    return res

# 递归版阶乘
def fact_recur(n):
    if not isinstance(n,int) or n < 0:
        raise ValueError("n必须是非负整数")
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n-1)

print("\n===== 阶乘测试 =====")
for num in [0,1,5,10]:
    print(f"n={num},迭代阶乘={fact_iter(num)},递归阶乘={fact_recur(num)}")

# 5. summary 返回最小、最大、平均值元组
def summary(values):
    if len(values) == 0:
        raise ValueError("列表不能为空")
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return min_v, max_v, avg_v

print("\n===== summary函数测试（元组解包接收结果） =====")
data = [85,58,92,76,60,89]
min_val, max_val, avg_val = summary(data) # 元组解包
print(f"最小值={min_val},最大值={max_val},平均值={avg_val:.2f}")

# 6. sum_all(*args)可变参数求和，测试：无参数、单个、多个参数
def sum_all(*args):
    return sum(args)

print("\n===== sum_all可变参数求和 =====")
print("无参数 sum_all() =", sum_all())
print("单个参数 sum_all(10) =", sum_all(10))
print("多个参数 sum_all(10,20,30,40) =", sum_all(10,20,30,40))
```

```python
# ========== 实验3：列表、字典、元组与基础统计 ==========
names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 1. zip()构建字典，items()输出
print("===== 1. zip构建字典，遍历items =====")
stu_dict = dict(zip(names, scores))
for name, score in stu_dict.items():
    print(f"{name}: {score}")

# 2. enumerate，序号从1开始
print("\n===== 2. enumerate输出序号、姓名、成绩 =====")
for idx, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"序号{idx} | {name} | {score}")

# 3. 列表推导式：提取及格姓名；成绩平方列表
print("\n===== 3. 列表推导式 =====")
pass_names = [name for name, s in zip(names, scores) if s >= 60]
score_square = [s**2 for s in scores]
print("及格学生姓名：", pass_names)
print("成绩平方列表：", score_square)

# 4. 统计指标：人数、平均分、最高分、最低分、及格率
print("\n===== 4. 基础统计指标 =====")
count = len(scores)
avg = sum(scores)/count
max_score = max(scores)
min_score = min(scores)
pass_cnt = len([s for s in scores if s >= 60])
pass_rate = pass_cnt / count
print(f"总人数：{count}")
print(f"平均分：{avg:.2f}")
print(f"最高分：{max_score}")
print(f"最低分：{min_score}")
print(f"及格率：{pass_rate:.2%}")

# 5. 列表引用alias与copy()拷贝对比
print("\n===== 5. 列表别名与浅拷贝对比 =====")
demo_list = [10,20,30,40]
alias = demo_list
copied = demo_list.copy()
print("修改前 demo_list:", demo_list)
alias[1] = 999
print("修改alias后 demo_list:", demo_list)
print("copied副本：", copied)
# 结论：alias是同一个列表的引用，改alias会影响原列表；copy()生成独立副本，互不影响

# 6. 元组不可变 + 长短不一序列zip测试
print("\n===== 6. 元组与不等长序列zip =====")
stu_tuple = ("Tom", 78)
print("元组记录：", stu_tuple)
# stu_tuple[1] = 90  # 取消注释运行会报错，元组元素不可赋值

list_a = [1,2,3]
list_b = ["x","y"]
zip_short = list(zip(list_a, list_b))
print("不等长序列zip结果，只匹配到最短长度：", zip_short)
```

```python
# ========== 实验4：文件读写、异常处理与模块 ==========
import os
import pickle

# 1. 创建存放数据的文件夹 exp01_data
data_dir = "exp01_data"
os.makedirs(data_dir, exist_ok=True)

# 定义解析单行学生记录的函数
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("一行必须包含姓名,成绩两个字段")
    name = parts[0].strip()
    try:
        score = int(parts[1].strip())
    except ValueError:
        raise ValueError("成绩必须是整数")
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在0~100范围内")
    return {"name": name, "score": score}


# 2. 写入 students.txt
students_txt_path = os.path.join(data_dir, "students.txt")
content = """Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
BadLine,abc
WrongScore,150
"""
with open(students_txt_path, "w", encoding="utf-8") as f:
    f.write(content)
print(f"✅ 已写入文件：{students_txt_path}")


# 3. 读取文件，try-except捕获每行错误，收集有效记录
valid_records = []
print("\n===== 读取并解析 students.txt =====")
with open(students_txt_path, "r", encoding="utf-8") as f:
    for line_no, line in enumerate(f, start=1):
        try:
            res = parse_student(line)
            if res is not None:
                valid_records.append(res)
                print(f"第{line_no}行 解析成功: {res}")
        except Exception as e:
            print(f"第{line_no}行 解析失败，错误：{e}")


# 4. 输出report.txt（w覆盖模式）
report_path = os.path.join(data_dir, "report.txt")
with open(report_path, "w", encoding="utf-8") as f:
    f.write("===== 学生成绩报告 =====\n")
    for rec in valid_records:
        f.write(f"{rec['name']} : {rec['score']}\n")
print(f"\n✅ report.txt（w覆盖模式）写入完成")

# a追加模式，追加一行记录
with open(report_path, "a", encoding="utf-8") as f:
    f.write("追加记录: Tom : 77\n")
print("✅ 使用a模式追加一行到 report.txt")


# 5. pickle保存与读取 students.pkl
pkl_path = os.path.join(data_dir, "students.pkl")
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)
print(f"\n✅ pickle加载出来的数据：{loaded_data}")


# 6. 将parse_student写入外部 student_utils.py 并导入使用
utils_path = os.path.join(data_dir, "student_utils.py")
utils_code = '''
def parse_student(line):
    line = line.strip()
    if not line:
        return None
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError("一行必须包含姓名,成绩两个字段")
    name = parts[0].strip()
    try:
        score = int(parts[1].strip())
    except ValueError:
        raise ValueError("成绩必须是整数")
    if not (0 <= score <= 100):
        raise ValueError("成绩必须在0~100范围内")
    return {"name": name, "score": score}
'''
with open(utils_path, "w", encoding="utf-8") as f:
    f.write(utils_code)

# 把exp01_data加入搜索路径，才能import
import sys
sys.path.insert(0, data_dir)
from student_utils import parse_student as imported_parse

test_line = "Lily,88"
test_result = imported_parse(test_line)
print(f"\n✅ 从外部py文件导入函数测试结果：{test_result}")


# 7. 捕获 FileNotFoundError，读取不存在的文件
print("\n===== 捕获文件不存在异常 =====")
nonexist_file = os.path.join(data_dir, "not_exist.txt")
try:
    with open(nonexist_file, "r", encoding="utf-8") as f:
        pass
except FileNotFoundError:
    print(f"捕获异常：文件 {nonexist_file} 不存在")
```

```python
# ========== 实验5：类、继承、对象复制 ==========
import copy

# 1. 定义Student基础类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        # 判断是否及格：>=60返回True
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, 及格={self.is_passed()})"


# 2. GraduateStudent继承Student，新增研究方向research_topic
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic}, 及格={self.is_passed()})"


# 3. 使用实验4得到的有效记录批量实例化Student对象
print("===== 批量创建Student对象 =====")
student_obj_list = []
# valid_records 是实验4解析得到的有效学生记录
for rec in valid_records:
    s_obj = Student(rec["name"], rec["score"])
    student_obj_list.append(s_obj)
    print(s_obj)


# 4. 测试研究生子类
print("\n===== 测试继承类 GraduateStudent =====")
gs = GraduateStudent("Tom",77,"大数据分析")
print(gs)


# 5. 浅拷贝 copy.copy() 和深拷贝 copy.deepcopy() 对比嵌套列表
print("\n===== 浅拷贝、深拷贝对比（嵌套列表） =====")
origin = [10, 20, [30, 40]]
shallow_cp = copy.copy(origin)
deep_cp = copy.deepcopy(origin)

print("修改前 origin =", origin)
shallow_cp[2][0] = 999
print("修改浅拷贝内部嵌套元素后：")
print("origin    : ", origin)
print("shallow_cp: ", shallow_cp)
print("deep_cp   : ", deep_cp)
# 结论：浅拷贝只复制外层，嵌套列表仍然共用同一个引用；深拷贝完全独立，互不影响
```
